Pydantic is used for data validation in python. They use Python type annotations and to define the structure and validate data at runtime .

```python
from pydantic import BaseModel
```

```python
class Person1(BaseModel):
    name: str
    city: str
    age: int

person1=Person1(name="Sachin",city="Ldh",age=21)
print(type(person1))
print(person1)
```

```python
from dataclasses import dataclass
@dataclass
class Person2():
    name: str
    city: str
    age: int

person2=Person2(name="Sachin",city="Ldh",age=21)
print(type(person2))
print(person2)
```

BaseModel will give error if values are not passed according to datatypes defined in class

```python
person3=Person1(name="Sachin",city=100,age=21)
print(person3)
```

But here no error will be there for dataclass

```python
person4=Person2(name="Sachin",city=100,age=21)
print(type(person4))
print(person4)
```

Optional Values

```python
from typing import Optional
class Employee(BaseModel):
    name:str
    age:int
    city: Optional[str]=None
    salary: Optional[float] =50000
employee=Employee(name='Sachin',age=21)
employee
```

here pydantic will automatically typecast salary into float

```python
employee1=Employee(name='Sachin',age=21,salary=50000)
employee1
```

```python
from typing import Optional
class classroom(BaseModel):
    name:list[str]
    std:int
    city: Optional[str]=None
classes=classroom(name=['Sachin','John','Starc'],std=7)
classes
```

```python
try:
    classes=classroom(name=['Sachin','John','Starc',111],std=7)
except Exception as e:
    print(e)
```

Nested Models

```python
#here also it type casted zip from str into int
class Address(BaseModel):
    street:str
    city:str
    zip:int
class Employee(BaseModel):
    name:str
    age:int
    address:Address
emp1=Employee(name='Sachin',age=21,address={"street":"123 Main Street","city":"London","zip":"11"})
```

Pydantic Fields : Customisation and Constraints

```python
from pydantic import BaseModel,Field
class Item(BaseModel):
    name:str=Field(min_length=5,max_length=20)
    price:float=Field(gt=100,le=1000)  #greater than 100 , less than 1000
item=Item(name="Dryer",price=599)
print(item)
```

Fields with default values

```python
class User(BaseModel):
    name:str=Field(...,description="Unique username to each user")  #... means it must be provide by the user
    age:int=Field(default=21,description="User age")
    email:str=Field(default_factory=lambda: "user@example.com",description="User email")
user=User(name="Sachin")
user
```
