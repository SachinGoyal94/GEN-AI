### HuggingFaceEndpoint
#### How to Access HuggingFace Models with API
There are also two ways to use this class. You can specify the model with the repo_id parameter. Those endpoints use the serverless API, which is particularly beneficial to people using pro accounts or enterprise hub. Still, regular users can already have access to a fair amount of request by connecting with their HF token in the environment where they are executing the code.

```python
from langchain_huggingface import HuggingFaceEndpoint
```

to save my local machine's hardware we create endpoints

so they connect to hugging face servers to get response

instead of installing heavy models and giving load to gpu

```python
os.environ["HUGGINGFACEHUB_API_TOKEN"] =os.getenv("HF_TOKEN")
#get these repo id from hugging face
repo_id="google/gemma-2-9b"
llm=HuggingFaceEndpoint(repo_id=repo_id,temperature=0.7)
llm
```

```python
llm.invoke("What is the power of machine learning??")
```

```python
repo_id="meta-llama/Llama-3.1-8B"
llm=HuggingFaceEndpoint(repo_id=repo_id,temperature=0.7)
llm.invoke("What is the power of Generative Ai ??")
```

```python
from langchain.prompts import PromptTemplate
prompt="you are an expert assistant,help user with their {Questions}"
prompt_template=PromptTemplate(input_variables=["Questions"],template=prompt)
prompt_template
```

```python
chain=prompt_template | llm
chain.invoke("What is the power of  BCCI in selecting the teams for cricket??")
```

```python
from langchain_community.embeddings import HuggingFaceBgeEmbeddings

model_name = "BAAI/bge-small-en"
model_kwargs = {"device": "cpu"}
encode_kwargs = {"normalize_embeddings": True}
hf = HuggingFaceBgeEmbeddings(
    model_name=model_name, model_kwargs=model_kwargs, encode_kwargs=encode_kwargs
)
hf
```

```python
hf.embed_query("hii this is an example of huggingface Bge embeddings")
```
